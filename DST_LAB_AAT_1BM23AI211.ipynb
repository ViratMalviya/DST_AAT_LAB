```python
def balance(a):
    brackets = {')': '(', '}': '{', ']': '['}
    empty_stack = []

    for value in a:
        if value in brackets.values():
            empty_stack.append(value)
        elif value in brackets.keys():
            if empty_stack and empty_stack[-1] == brackets[value]:
                empty_stack.pop()
            else:
                return "NO"
    return "YES" if not empty_stack else "NO"

if __name__ == '__main__':
    bracket = int(input().strip())
    for i in range(bracket):
        a = input()
        answer = balance(a)
        print(answer)
```

```python
def twoStacks(maxSum, a, b):
    score = 0
    totalSum = 0

    for x in a:
        if totalSum + x <= maxSum:
            totalSum += x
            score += 1
        else:
            break

    maxScore = score
    currScore = score

    for i, y in enumerate(b):
        totalSum += y
        currScore += 1

        while totalSum > maxSum and score > 0:
            totalSum -= a[score - 1]
            score -= 1

        if totalSum <= maxSum:
            maxScore = max(maxScore, currScore)
        else:
            break

    return maxScore

maxSum = 12
a = [4, 2, 4, 6, 1]
b = [2, 1, 8, 5]

result = twoStacks(maxSum, a, b)
print(result)
```

```python
class Stackqueue:
    def __init__(self):
        self.stack1 = []
        self.stack2 = []

    def enqueue(self, value):
        self.stack1.append(value)

    def dequeue(self):
        if not self.stack2:
            while self.stack1:
                self.stack2.append(self.stack1.pop())
        if not self.stack2:
            raise IndexError("Queue is empty!")
        return self.stack2.pop()

    def peek(self):
        if not self.stack2:
            while self.stack1:
                self.stack2.append(self.stack1.pop())
        if not self.stack2:
            raise IndexError("Queue is empty!")
        return self.stack2[-1]


def process_queries(queries):
    queue = Stackqueue()
    results = []

    for query in queries:
        if query[0] == 1:
            queue.enqueue(query[1])
        elif query[0] == 2:
            queue.dequeue()
        elif query[0] == 3:
            results.append(queue.peek())

    return results

queries = [
    [1, 10],
    [1, 20],
    [3],
    [2],
    [3],
    [1, 30],
    [2],
    [3],
    [1, 40],
    [3]
]

output = process_queries(queries)
for result in output:
    print(result)
```
